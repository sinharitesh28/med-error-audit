```python
# [Jupyter Notebook Script]
import re
from pathlib import Path

cwd = Path.cwd()
login_html_file = cwd / "frontend" / "login.html"

if login_html_file.exists():
    content = login_html_file.read_text(errors='ignore')
    
    # 1. Fix the strange encoding characters
    content = content.replace("— OR —", "- OR -")
    content = content.replace("Ã¢â‚¬â€ OR Ã¢â‚¬â€", "- OR -")
    
    # 2. Add Cache-Busting and No-Store headers to the Polling Fetch
    # Find the dynamic fetch (whether it has localhost or relative path)
    old_fetch_pattern = r'const pollRes = await fetch\(`.*?/api/auth/poll-session/\$\{token\}`\);'
    
    new_fetch = """// Force the browser to skip cache and hit the live database
                        const pollRes = await fetch(`/api/auth/poll-session/${token}?_t=` + new Date().getTime(), {
                            method: 'GET',
                            headers: {
                                'Cache-Control': 'no-cache, no-store, must-revalidate',
                                'Pragma': 'no-cache',
                                'Expires': '0'
                            },
                            cache: 'no-store'
                        });"""
                        
    content = re.sub(old_fetch_pattern, new_fetch, content)
    
    login_html_file.write_text(content, encoding='utf-8')
    print(" [X] Successfully applied Cache-Busting and Encoding fixes to login.html")
else:
    print(" [!] login.html not found. Check if your Jupyter is in the correct root folder.")
```
